# Análise Exploratória de Dados Automática
Este notebook itera sobre todos os ficheiros CSV presentes em `data/processed/` e gera uma análise exploratória rápida para cada um, tirando partido das capacidades visuais do Jupyter.

```python
import os
import pandas as pd
from IPython.display import display, Markdown
```

```python
current_dir = os.getcwd()
root_dir = os.path.dirname(current_dir) if os.path.basename(current_dir) == 'notebooks' else current_dir
data_dir = os.path.join(root_dir, 'data', 'processed')

if not os.path.exists(data_dir):
    print(f"Error: Directory not found {data_dir}")
    print("Execute 'src/data_prep.py' script to generate data first.")
    csv_files = []
else:
    csv_files = [f for f in os.listdir(data_dir) if f.endswith('.csv')]
    print(f"Found {len(csv_files)} datasets. Initiating analysis...")
```

```python
for file in csv_files:
    file_path = os.path.join(data_dir, file)
    dataset_name = os.path.splitext(file)[0]
    
    display(Markdown(f"<br>\n\n---\n# Dataset: **{dataset_name}**"))
    
    try:
        df = pd.read_csv(file_path)
        
        display(Markdown(f"### General Info\n- **Number of Rows:** {df.shape[0]}\n- **Number of Columns:** {df.shape[1]}"))
        
        display(Markdown("### Data Types and Non-Null Counts"))
        df.info()
        
        display(Markdown("### Data Sample (First 5 Rows)"))
        display(df.head())
        
        display(Markdown("### Descriptive Statistics"))
        display(df.describe(include='all'))
        
        display(Markdown("### Missing Values"))
        nulos = df.isnull().sum()
        display(nulos[nulos > 0].to_frame(name='Null Values') if nulos.sum() > 0 else "None found.")
        
    except Exception as e:
        print(f"Error processing {file}: {e}")
```
