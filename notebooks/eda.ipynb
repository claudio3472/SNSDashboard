# Análise Exploratória de Dados Automática
Este notebook itera sobre todos os ficheiros CSV presentes em `data/processed/` e gera uma análise exploratória rápida para cada um, tirando partido das capacidades visuais do Jupyter.

```python
import os
import pandas as pd
from IPython.display import display, Markdown
```

```python
current_dir = os.getcwd()
root_dir = os.path.dirname(current_dir) if os.path.basename(current_dir) == 'notebooks' else current_dir
data_dir = os.path.join(root_dir, 'data', 'processed')

if not os.path.exists(data_dir):
    print(f"Error: Directory not found {data_dir}")
    print("Execute 'src/data_prep.py' script to generate data first.")
    csv_files = []
else:
    csv_files = [f for f in os.listdir(data_dir) if f.endswith('.csv')]
    print(f"Found {len(csv_files)} datasets. Initiating analysis...")
```

```python
for file in csv_files:
    file_path = os.path.join(data_dir, file)
    dataset_name = os.path.splitext(file)[0]
    
    display(Markdown(f"<br>\n\n---\n# Dataset: **{dataset_name}**"))
    
    try:
        df = pd.read_csv(file_path)
        
        display(Markdown(f"### General Info\n- **Number of Rows:** {df.shape[0]}\n- **Number of Columns:** {df.shape[1]}"))
        
        display(Markdown("### Data Types and Non-Null Counts"))
        df.info()
        
        display(Markdown("### Data Sample (First 5 Rows)"))
        display(df.head())
        
        display(Markdown("### Descriptive Statistics"))
        display(df.describe(include='all'))
        
        display(Markdown("### Missing Values"))
        nulos = df.isnull().sum()
        display(nulos[nulos > 0].to_frame(name='Null Values') if nulos.sum() > 0 else "None found.")
        
        # Adicionar visualização dos valores únicos da coluna 'entidade'
        if 'entidade' in df.columns:
            display(Markdown("### Unique Values in 'entidade'"))
            unique_entidades = sorted(df['entidade'].unique())
            display(Markdown("\n".join(f"- {val}" for val in unique_entidades)))
        else:
            display(Markdown("### Unique Values in 'entidade'\nColumn 'entidade' not found in this dataset."))
        
        # Adicionar visualização dos valores únicos da coluna 'instituicao'
        if 'instituicao' in df.columns:
            display(Markdown("### Unique Values in 'instituicao'"))
            unique_instituicoes = sorted(df['instituicao'].unique())
            display(Markdown("\n".join(f"- {val}" for val in unique_instituicoes)))
        else:
            display(Markdown("### Unique Values in 'instituicao'\nColumn 'instituicao' not found in this dataset."))
        
        # Mostrar instituições/entidades por região
        if 'regiao' in df.columns:
            col = 'entidade' if 'entidade' in df.columns else ('instituicao' if 'instituicao' in df.columns else None)
            if col:
                display(Markdown(f"### {col.capitalize()} by Region"))
                grouped = df.groupby('regiao')[col].unique()
                for regiao, vals in grouped.items():
                    display(Markdown(f"**Region: {regiao}**"))
                    display(Markdown("\n".join(f"- {val}" for val in sorted(vals))))
            else:
                display(Markdown("### Entities/Institutions by Region\nNo 'entidade' or 'instituicao' column found."))
        else:
            display(Markdown("### Entities/Institutions by Region\nColumn 'regiao' not found in this dataset."))
        
    except Exception as e:
        print(f"Error processing {file}: {e}")
```
