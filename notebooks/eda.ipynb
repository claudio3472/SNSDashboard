```python
import os
import pandas as pd
import plotly.express as px

# Find processed data folder.
root_dir = os.path.dirname(os.getcwd())
processed_dir = os.path.join(root_dir, "data", "processed")

# List all CSV files.
files = [f for f in os.listdir(processed_dir) if f.endswith('.csv')]

print(f"Analysing {len(files)} datasets...\n")

for file in files:
    full_path = os.path.join(processed_dir, file)
    dataset_name = file.replace('.csv', '')
    
    print(f"\n{'='*60}")
    print(f"Dataset: {dataset_name.upper()}")
    print(f"{'='*60}")
    
    df = pd.read_csv(full_path)
    
    print(f"Dimension: {df.shape[0]:,} rows x {df.shape[1]} columns\n")
    
    # Null values.
    null_values = df.isna().sum()
    relevant_null_values = null_values[null_values > 0]
    
    if not relevant_null_values.empty:
        print("Top 5 null values found:")
        print(relevant_null_values.sort_values(ascending=False).head(5))
        # Show the percentage of null_values for the worst column.
        worst_col = relevant_null_values.idxmax()
        perc = (relevant_null_values.max() / len(df)) * 100
        print(f"   -> The column '{worst_col}' has {perc:.1f}% missing values.")
    else:
        print("No null values found.")
        
    print("\n" + "-"*60 + "\n")
    
    # Anual distribution. See if there are any years with incomplete data.
    if 'ano' in df.columns:
        anual_count = df['ano'].value_counts().reset_index()
        anual_count.columns = ['ano', 'contagem']
        anual_count = anual_count.sort_values('ano')
        
        # Only show full years.
        anual_count['ano'] = anual_count['ano'].astype(int).astype(str) 
        
        fig = px.bar(
            anual_count, 
            x='ano', 
            y='contagem', 
            title=f"Data volume per year - {dataset_name}",
            labels={'ano': 'Year', 'contagem': 'Number of records'},
            template="plotly_white",
            color_discrete_sequence=['#3498db'],
            height=350
        )
        fig.show()
    else:
        print("This dataset doesn't have the column 'ano' to plot time evolution.")
```
